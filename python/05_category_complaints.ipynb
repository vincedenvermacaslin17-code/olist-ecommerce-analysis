```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
category_complaints_revenue = pd.read_csv("../results/05_category_complaints.csv")
print(category_complaints_revenue.shape)
category_complaints_revenue.head()
```

```python
category_complaints_revenue.plot(x="revenue", y="complaint_rate", kind="scatter")
plt.title("Revenue vs complaint rate by product category")
plt.xlabel("Revenue (BRL)")
plt.ylabel("Complaint rate (%)")
plt.show()

```

```python
category_complaints_revenue["revenue"].corr(category_complaints_revenue["complaint_rate"])
```

**Findings**

The chart shows a cloud of dots with no specific pattern. The correlation between revenue and complaint rate is 0.048, which means revenue doesn't predict complaint rate. Categories with high revenue do not get fewer complaints. The data shows only revenue, not profit, because it has no data on sellers' costs.
