```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
customer_types = pd.read_csv("../results/04_customer_types.csv")
print(customer_types.shape)
customer_types["share"] = (customer_types["revenue"] / customer_types["revenue"].sum()) * 100
print(customer_types)
```

```python
customer = customer_types.plot(x="customer_type", y="share", kind="bar", legend=False)
customer.bar_label(customer.containers[0], fmt="%.2f")
plt.title("Customer Types")
plt.ylabel("Revenue Share Percentage")
plt.xlabel("Customer Types")
plt.show()
```

**Findings**

- One-time customers bring in 94.4% of delivered revenue, repeat only 5.6%
- Olist doesn't depend on returning customers
- There is a massive percentage gap between the two customer types ( one-time and repeat customer) but that does not consider how many customers are in each group.
