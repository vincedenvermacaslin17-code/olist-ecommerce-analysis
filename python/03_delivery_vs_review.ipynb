```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
delivery_vs_review = pd.read_csv("../results/03_delivery_vs_review.csv")
```

```python
delivery_vs_review.head()
```

```python
delivery_vs_review.plot(x="review_score", y="delivery_days", kind="bar", legend=False)
plt.title("Delivery Days vs Review Score")
plt.xlabel("Review Scores")
plt.ylabel("Delivery Days")
plt.show()
```

**Findings**

The chart shows the average delivery days for each review score. Orders with lower scores had longer delivery times on average. This doesn't prove that slow delivery causes low scores. Customer location may affect both delivery days and review scores.
